## Hierarchical Inheritance

Hierarchical inheritance means multiple child classes inherit from the same parent class.

```python
from abc import ABC,abstractmethod

class Shape(ABC):

  @abstractmethod

  def area(self):

    pass

class Circle(Shape):

  def area(self):

    print("Calculating circle area")

obj=Circle()

obj.area()
```

## Special Methods: __len__

The __len__ method defines what len(object) returns.

### Example 1

```python
class Library:

  def __init__(self,items):

    self.items=items

  def __len__(self):

    return len(self.items)

library=Library((['Python','Javea','C']))

print(len(library))
```

### Example 3

## Operator Overloading: __add__

The __add__ method customizes what the + operator does for objects of a class.

### Example 1

```python
class Number:

  def __init__(self,value):

    self.value=value

  def __add__(self,other):

    return self.value+other.value

n1=Number(10)

n2=Number(20)

print(n1+n2)
```

### Example 2

```python
class Number:

  def __init__(self,value):

    self.value=value

  def __add__(self,other):

    return Number(self.value+other.value)

n1=Number(15)

n2=Number(25)

result=n1+n2

print(result.value)
```

### Example 3

```python
class Money:

  def __init__(self,amount):

    self.amount=amount

  def __add__(self,other):

    return Money(self.amount+other.amount)

m1=Money(500)

m2=Money(250)

total=m1+m2

print(total.amount)
```

## Quick Review

- **Class:** A blueprint for creating objects.
- **Object:** An instance of a class.
- **`self`:** Refers to the current instance inside instance methods.
- **`__init__`:** Initializes an object's attributes.
- **Inheritance:** Allows a child class to reuse parent-class behavior.
- **`super()`:** Accesses parent-class behavior according to Python's method resolution order.
- **Class attribute:** Defined on the class and shared unless shadowed by an instance attribute.
- **`@classmethod`:** Receives the class as `cls`.
- **`@staticmethod`:** Does not automatically receive `self` or `cls`.
- **Polymorphism:** Different classes can respond to the same method call differently.
- **Abstraction:** Defines required behavior through abstract methods.
- **Encapsulation:** Keeps implementation details behind a class interface.
- **Composition:** Builds a class using objects of other classes.
- **Special methods:** Let objects work with built-in operations such as `print()`, `len()`, and `+`.
